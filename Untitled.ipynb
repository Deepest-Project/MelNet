```python
from torch.utils.data import Dataset, DataLoader
import scipy, librosa
from audio_processing import *
import os
import torch
import hparams

class Spliter():
    def __init__(self):
        self.split=0
    
    def reset(self):
        self.split=0
        
    def __call__(self, melspec):
        if self.split%2==0:
            self.split += 1
            return melspec[0::2, :], melspec[1::2, :]
        
        elif self.split%2==1:
            self.split += 1
            return melspec[:, 0::2], melspec[:, 1::2]

def pad_mel(melspecs):
    B, F, T = len(melspecs), melspecs[0].shape[0], max([x.shape[1] for x in melspecs])
    padded_mel =  np.zeros((B, F, T))
    for i, mel in enumerate(melspecs):
        padded_mel[i, :, :mel.shape[1]] = mel

    return padded_mel


class MelData(Dataset):
    def __init__(self, hparams):
        super(Dataset, self).__init__()
        self.root_dir = hparams.root_dir
        self.n_tiers = hparams.n_tiers
        self.sr = hparams.sr
        self.n_fft = hparams.n_fft
        self.n_mels = hparams.n_mels
        self.n_hop = hparams.n_hop
        self.n_overlap = hparams.n_fft - hparams.n_hop
        
        self.wav_files = list(filter(lambda f: f.endswith('.wav'), os.listdir(self.root_dir)))
        self.split = Spliter()
        self.mel_basis = librosa.filters.mel(hparams.sr, hparams.n_fft, hparams.n_mels)
        
    def __getitem__(self, i):
        ##### Wav -> Melspectrogram #####
        wav, _ = librosa.load( os.path.join(self.root_dir, self.wav_files[i]), sr=22050)
        wav = normalize(wav)
        wav = trim(wav)
        melspec = wav_to_melspec(wav, self.sr, self.n_fft, self.n_hop, self.n_mels, self.mel_basis)
        
        ##### Melspectrogram Validation #####
        n_half_t = (self.n_tiers-1) // 2
        n_time = melspec.shape[1] - melspec.shape[1] % 2**n_half_t
        melspec = melspec[:, :n_time]
        
        ##### Build mel_tiers #####
        mel_tiers = ['DUMMY_TIER'] + [ 0 for _ in range(self.n_tiers) ]
        for t in range(self.n_tiers, 1, -1):
            tier, melspec = self.split(melspec)
            mel_tiers[t] = tier
        mel_tiers[1] = melspec
        self.split.reset()
        
        return mel_tiers
    
    def __len__(self):
        return len(self.wav_files)
    
class MelCollate():
    def __init__(self, hparams):
        self.n_tiers = hparams.n_tiers
        
    def __call__(self, batch):
        mel_tiers = ['DUMMY_TIER'] + [ [] for _ in range(self.n_tiers) ]
        
        for data in batch:
            for t in range(1, self.n_tiers+1):
                mel_tiers[t].append(data[t])
        
        for t in range(1, self.n_tiers+1):
            mel_tiers[t] = pad_mel(mel_tiers[t])
        
        return mel_tiers
```

```python
dataset = MelData(hparams)
collate_fn = MelCollate(hparams)
dataloader = DataLoader(dataset, shuffle=True, batch_size=32, drop_last=True, collate_fn=collate_fn)
mel_tiers = next(iter(dataloader))
```
